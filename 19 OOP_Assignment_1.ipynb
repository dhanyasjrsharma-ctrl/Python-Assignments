## Question 2

Create a `Student` class using constructor to store name, age and course. Create two objects and display details.

**Sample Input**
```
Use suitable values
```

**Expected Output**
```
Display the required details correctly.
```

````python
class Student():
    def __init__(self,name,age,course): #__init__() is a special method that initialises an object, so it is what we are saying "create a class using constructor" in que
        self.name = name
        self.age = age
        self.course = course
    def show_details(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)
s1 = Student("Dhanya",19,"Btech")
s1.show_details()

````

## Question 3

Create a `BagFactory` class with constructor accepting material, zips and pockets. Create two bag objects and print details.

**Sample Input**

Use suitable values

**Expected Output**
