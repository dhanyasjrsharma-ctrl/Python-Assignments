## Q1. Student Class

Create a class `Student` with name, age and course. Create an object and display all details.

```
Sample Input:
Name = Rahul
Age = 20
Course = Python

Sample Output:
Name : Rahul
Age : 20
Course : Python
```

````python
class Student:
    def __init__(self,name,age,course):
        self.name = name
        self.age = age
        self.course = course
    def showdetails(self): # method
        print("Name : ",self.name)
        print("Age : ",self.age)
        print("Course : ",self.course)
s1 = Student("Rahul",20,"Python")
s1.showdetails()
````

## Q10. Circle Class

Create a class `Circle` with radius. Create method `area()`. Use 3.14*r*r.

```
Sample Input:
Radius = 7

Sample Output:
Area = 153.86
```

````python
class Circle:
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return self.radius ** 2 * 3.14 # we can write 22/7 also for pi
a1 = Circle(7)
a1.area()
````
